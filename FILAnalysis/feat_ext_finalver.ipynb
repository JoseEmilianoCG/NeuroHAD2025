```python
import preprocessing as prep
import processing as proc
import json
import os
import pickle
import glob
```

```python
filter_params = {
    "filter_order": 2,
    "bandwidth": [4, 50],
    "type": "bandpass",
    "fs": 256,
    "output": "sos",
}

proc_config = {
    "sessions": [2, 12, 14, 21, 27, 28, 31, 39, 41, 42],
    "stages": ["baseline", "shared", "individual"],
    "min_freq": 4,
    "max_freq": 30,
    "channels": ["TP9", "TP10", "AF7", "AF8"],
    "bandfreqs": {
        "theta": [4, 8],
        "alpha": [8, 13],
        "beta": [13, 30],
        # "gamma": [30, 50],
    },
}

window_config = {"fs": 256, "winsz": 4, "ovlap": 0.5}
```

```python
cwd = os.getcwd()

preprodata = prep.prepro(
    stages=proc_config["stages"], sessions=proc_config["sessions"], cwd=cwd, filter_params=filter_params
)
```

```python
preprodata['p1']['shared']
```

```python
procdata = proc.process(
            preprodata=preprodata,
            stages=proc_config["stages"],
            sessions=proc_config["sessions"],
            min_freq=proc_config["min_freq"],
            max_freq=proc_config["max_freq"],
            fs=window_config["fs"],
            winsz=window_config["winsz"],
            ovlap=window_config["ovlap"],
            channels=proc_config["channels"],
        )
```

```python
procdata['baseline']
```

```python
band_procdata = proc.get_band_average(
    procdata=procdata,
    band_freqs=proc_config["bandfreqs"],
    stages=proc_config["stages"],
    channels=proc_config["channels"],
)
```

```python
band_procdata['baseline']
```

```python
for state, df in band_procdata.items():
    df.to_csv(f'band_conndata_{state}.csv', index=False)
```
